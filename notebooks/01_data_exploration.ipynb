```python
def recommend_for_user(user_id, k=5):
    if user_id not in user_to_index:
        return []

    user_index = user_to_index[user_id]
    target_user_row = user_item_matrix[user_index]

    # Users with at least 5 interactions
    user_activity = user_item_matrix.getnnz(axis=1)
    eligible_users = np.where(user_activity >= 5)[0]

    # Remove target user from eligible users
    eligible_users = eligible_users[eligible_users != user_index]

    # Compare target user with eligible users
    eligible_user_matrix = user_item_matrix[eligible_users]

    similarity_scores = cosine_similarity(
        target_user_row,
        eligible_user_matrix
    )[0]

    # Top similar users
    top_indices = similarity_scores.argsort()[::-1][:10]
    similar_users = eligible_users[top_indices]

    # Products already seen by target user
    user_items = set(target_user_row.indices)

    # Similarity-weighted candidate scores
    candidate_scores = {}

    for position, user in enumerate(similar_users):
        similarity_score = similarity_scores[top_indices[position]]
        user_row = user_item_matrix[user]

        for item_index in user_row.indices:
            if item_index not in user_items:
                candidate_scores[item_index] = (
                    candidate_scores.get(item_index, 0)
                    + similarity_score
                )

    # Rank candidates
    ranked_candidates = sorted(
        candidate_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    # Top-K
    top_candidates = ranked_candidates[:k]

    # Convert matrix indexes to real product IDs
    recommendations = [
        (index_to_item[item_index], score)
        for item_index, score in top_candidates
    ]

    return recommendations
```

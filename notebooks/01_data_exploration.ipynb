## Building Sparse Matrix

```python
from scipy.sparse import csr_matrix
```

```python
user_item_matrix = csr_matrix(
    (
        user_item["interaction_score"],
        (user_item["user_index"], user_item["item_index"])
    )
)
```

```python
user_item_matrix.shape ## (users,products)
```

```python
index_to_item = {
    index: item_id
    for item_id, index in item_to_index.items()
}

similar_item_ids = [
    index_to_item[i]
    for i in similar_user_row.indices
]

similar_item_ids
```

```python
candidate_items = []
```

```python
for user in similar_users:
    user_row = user_item_matrix[user]
    
    for item_index in user_row.indices:
        candidate_items.append(item_index)
```

```python
similar_user_scores = {
    user: user_similarity[0][user]
    for user in similar_users
}

similar_user_scores
```

```python
candidate_scores = {}
for user, similarity_score in similar_user_scores.items():
    user_row = user_item_matrix[user]

    for item_index in user_row.indices:
        if item_index not in user_items:
            candidate_scores[item_index] = (
                candidate_scores.get(item_index, 0) + similarity_score
            )
candidate_scores
```

```python
ranked_candidates = sorted(
    candidate_scores.items(),
    key=lambda x: x[1],
    reverse=True
)

ranked_candidates
```

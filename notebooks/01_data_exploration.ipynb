```python
all_items = pd.concat([items, items2], ignore_index=True)
```

```python
interacted_items = df["itemid"].unique()
```

```python
product_items = all_items[all_items["itemid"].isin(interacted_items)]
```

```python
product_items.shape
```

```python
product_items[["itemid", "property", "value"]].duplicated().sum()
```

```python
product_features = product_items[["itemid", "property", "value"]].drop_duplicates()
```

```python
product_features.shape
```

```python
product_features.groupby("itemid")
```

```python
product_features.head(10)
```

```python
product_features["feature"] = (
    product_features["property"].astype(str)
    + "_"
    + product_features["value"].astype(str)
)
```

```python
product_features.head(10)
```

```python
product_features["feature"] = (
    product_features["property"].astype(str)
    + "_"
    + product_features["value"].astype(str)
)
```

```python
product_features.head(10)
```

```python
product_profiles = (
    product_features.groupby("itemid")["feature"]
    .apply(" ".join)
)
```

```python
product_profiles.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(product_profiles)
tfidf_matrix.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(tfidf_matrix[0],tfidf_matrix)
similarity.shape
```

```python
similarity[0].argsort()
```

```python
similarity[0].argsort()[::-1]
```

```python
similar_products = similarity[0].argsort()[::-1][:11]

similar_products
```

```python
product_profiles.index[0]
```

```python
recommended_itemids = product_profiles.index[similar_products]

recommended_itemids
```

```python
similarity_scores = similarity[0][similar_products]

similarity_scores
```

```python
recommended_itemids = recommended_itemids[1:]
similarity_scores = similarity_scores[1:]
```

```python
recommended_itemids
```

```python
recommendations = pd.DataFrame({
    "itemid": recommended_itemids,
    "similarity": similarity_scores
})

recommendations
```
